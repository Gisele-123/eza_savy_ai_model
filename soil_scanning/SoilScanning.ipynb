```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('final_soil_dataset.csv')

print(df.head())

print(df.info())

print("Unique Soil Categories:", df['Soil_Category'].unique())
print("Unique Plants:", df['Plant_Name'].nunique())
```

```python
print(df.isnull().sum())

df = df.dropna(subset=['Soil_Category', 'Nitrogen', 'Phosphorus', 'Potassium', 'Plant_Name'])

df.fillna({
    'pH': df['pH'].median(),
    'Light_Hours': df['Light_Hours'].median(),
    'Season': df['Season'].mode()[0]
}, inplace=True)
```
